```python
import joblib
```

```python
results = joblib.load("/tmp/test_list.pkl")
```

```python
assert len(results) == 7, f"{len(results)}"
```

```python
for k, result in results.items():
    print("=" * 80)
    print("question:", result["question"])
    print("-" * 80)
    print("answer:", result["answer"])
    print("-" * 80)
    for chunk_item in result["aug_log"]["result"]:
        print("." * 80)
        print(chunk_item)
    print("." * 80)
```

```python
max_idx = len(results) - 1

for i, (k, result) in enumerate(results.items()):
    if i == max_idx:
        break
    assert result["aug_log"]["result"] == results[i+1]["aug_log"]["result"], f"i: {i}"
```
