```python
import os
import json
from utils import DataNormalizer, DataReader, DataTranslator
```

```python
current_dir = os.getcwd()
exported_to = current_dir + "/" + "normalized-sample.jsonl"
dataset_source = "/home/felipe/dev/projects/judiciary-system/notebooks/data/output.jsonl"
```

```python
def export_normalized_data(source_file_path: str, export_file_path: str, n_records: int = 0) -> None:
    """Exports the normalized data to a JSONL file."""
    reader = DataReader(source_file_path)
    normalizer = DataNormalizer()
    translator = DataTranslator()
    with open(export_file_path, "w", encoding="utf-8") as f:
        for line, doc in reader.read_data():
            if n_records > 0 and line >= n_records:
                break
            doc =  translator.translate_document_fields(doc)
            doc = normalizer.normalize_document(doc)
            record = json.dumps(doc, ensure_ascii=False)
            f.write(record + '\n')
```

```python
# n_records is the number of records normalized/exported. If ignored or 0 is given, it will normalize the whole data in the source file
export_normalized_data(source_file_path=dataset_source, export_file_path=exported_to, n_records=1)
```
