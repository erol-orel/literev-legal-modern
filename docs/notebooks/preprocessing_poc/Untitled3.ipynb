```python
import joblib
```

```python
list_stopwords = joblib.load("common_words")
```

```python
type(list_stopwords)
```

```python
n_stopwords = list(list_stopwords)
n_stopwords.sort()
with open("stop_words_from_tfidfVect.txt", "w") as f:
    for word in n_stopwords:
        f.write(word)
        f.write("\n")
```
