```python
# update stopwords
with open("stopwords.txt", "r") as f:
    reader = f.read()
words = reader.splitlines()
```

```python
len(words)
```

```python
words[-1]
```

```python
new_set_words = set()
for word in words:
    new_set_words.add(word.strip())
```

```python
len(new_set_words)
list_new_words = list(new_set_words)
list_new_words.sort()
```

```python
with open("new_stopwords.txt", "w") as f:
    for word in list_new_words:
        f.write(word)
        f.write("\n")
```
