```python
corpus = "Refus de supprimer du dossier de police de la recourante un rapport de renseignements et ses annexes. Déjà condamnée, il n'y avait pas lieu de prendre en compte l'aspect de répression des infractions. Condamnée pour voies de faits : contravention qui ne rentre pas dans le but de prévention des crimes et délits prévu par la LCBVM. Examen de l'ensemble des circonstances (ordonnance pénale et de non-entrée en matière partielle ne figure pas dans les dossiers de police et n'y est pas mentionnée ; condamnation pour une contravention, absence d'antécédents, absence de gravité comme la criminalité organisée ou les crimes et délits contre l'intégrité physique et sexuelle, moins d'un an depuis le rapport de renseignements et la condamnation). Absence d'intérêt à la conservation des documents litigieux. Recours admis et radiation du document ordonnée."
```

```python
from lingua import Language, LanguageDetectorBuilder
```

```python
SUPPORTED_LANGUAGES = [Language.ENGLISH, Language.FRENCH]
detector = LanguageDetectorBuilder.from_languages(*SUPPORTED_LANGUAGES).build()
```

```python
detected_language = detector.detect_language_of(corpus)

    # Check if the detected language is English
print(detected_language == Language.FRENCH)
```
