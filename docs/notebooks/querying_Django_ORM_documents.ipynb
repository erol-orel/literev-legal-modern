# Using Django's ORM (Object-Relational Mapping)

## Counts the total number of documents by project

````python
import asyncio
from asgiref.sync import sync_to_async
from literev.models import Document
from django.db.models import Count

@sync_to_async
def count_documents_by_project():
    document_counts = Document.objects.values('project__id', 'project__name').annotate(total=Count('id')).order_by('project__id')
    return [(count['project__id'], count['project__name'], count['total']) for count in document_counts]

async def main():
    document_counts = await count_documents_by_project()
    for project_id, project_name, total in document_counts:
        print(f"Project ID: {project_id}, Project Name: {project_name}, Total Documents: {total}")

# For Jupyter Notebook or an environment with an already running event loop
loop = asyncio.get_event_loop()
if loop.is_running():
    # Scheduling the coroutine to be run on the existing event loop
    task = loop.create_task(main())
else:
    # This is for environments without a running event loop
    asyncio.run(main())
````

## Query all data from the Document model including the project and decision_type fields, 

This code snippet queries all instances of the Document model and iterates over them to print out all required fields, along with the decision_type of each document.

**Fields(table)**

```
Proceduretype = A/491/2023
decisiontype = ATA/519/2023
decisiondate = 22.05.2023
result = REJETE
descriptors = Null
standard
```


````python
# Import the necessary function
from asgiref.sync import sync_to_async
from literev.models import Document

# Define an asynchronous function to fetch documents
@sync_to_async
def get_documents():
    documents = Document.objects.all()
    return [
        (
            document.project.id,
            document.project.name,
            document.procedure_type,
            document.decision_type,
            document.decision_date,
            document.descriptors,
            document.result,
            document.standards
        ) for document in documents]

# Use the async function in an asynchronous context
import asyncio
Q
async def main():
    documents = await get_documents()
    for project_id, project_name, procedure_type, decision_type, decision_date, descriptors, result, standards in documents:
        print(f"""
            Project ID: {project_id}, 
            Project Name: {project_name},
            Procedure_type: {procedure_type},
            Decision Type: {decision_type}, 
            Decision Date: {decision_date},
            Descriptors: {descriptors},
            Result: {result},
            Standards:  {standards}
            """)

# Run the async function
await main()
````
